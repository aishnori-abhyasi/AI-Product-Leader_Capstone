```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# 1. Generate 90 days of synthetic UPI transactions for a gig worker
np.random.seed(42)
start_date = datetime.now() - timedelta(days=90)
dates = [start_date + timedelta(days=i) for i in range(90) for _ in range(np.random.randint(1, 4))]

data = []
for date in dates:
    txn_type = np.random.choice(['Inflow', 'Outflow'], p=[0.3, 0.7])
    if txn_type == 'Inflow':
        # Simulating gig platform payouts vs personal transfers
        category = np.random.choice(['Platform_Payout', 'Peer_Transfer'], p=[0.7, 0.3])
        amount = np.random.exponential(scale=1500) if category == 'Platform_Payout' else np.random.exponential(scale=500)
    else:
        category = np.random.choice(['Rent_Utilities', 'Groceries', 'Fuel', 'Leisure'], p=[0.2, 0.4, 0.2, 0.2])
        amount = np.random.exponential(scale=300)

    data.append({
        'Timestamp': date.strftime('%Y-%m-%d %H:%M:%S'),
        'Type': txn_type,
        'Category': category,
        'Amount': round(amount, 2)
    })

df = pd.DataFrame(data)
```

```python
# 2. AI Product Leader: Structural Analytics: Extract core operational risk metrics
total_inflow = df[df['Type'] == 'Inflow']['Amount'].sum()
total_outflow = df[df['Type'] == 'Outflow']['Amount'].sum()
net_savings_ratio = (total_inflow - total_outflow) / total_inflow if total_inflow > 0 else 0
platform_reliance = df[df['Category'] == 'Platform_Payout']['Amount'].sum() / total_inflow if total_inflow > 0 else 0

print(f"--- 90-Day UPI Telemetry Summary ---")
print(f"Total Operational Inflow: ₹{total_inflow:,.2f}")
print(f"Total Operational Outflow: ₹{total_outflow:,.2f}")
print(f"Net Savings Margin: {net_savings_ratio*100:.2f}%")
print(f"Platform Revenue Dependency (Inflow Moat): {platform_reliance*100:.2f}%\n")
```

```python
# 3. Simple Rule-Based Credit Tiering Baseline
if net_savings_ratio > 0.15 and platform_reliance > 0.50:
    print("Pre-Approval Decision: TIER 1 (Low Risk) -> Eligible for Micro-Investment matching + low-interest micro-loans.")
elif net_savings_ratio > 0.05:
    print("Pre-Approval Decision: TIER 2 (Moderate Risk) -> Eligible for structured micro-loans with milestone parameters.")
else:
    print("Pre-Approval Decision: TIER 3 (High Risk) -> Route strictly to savings-builder financial coaching module.")
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# 1. Setup Configuration for 10,000 Gig Workers
num_workers = 10000
np.random.seed(42)

# Define demographic archetypes with unique financial signatures
archetypes = {
    'Kirana_Owner':        {'inflow_avg': 4500,  'inflow_count': 12, 'outflow_ratio': 0.85, 'stability': 0.90},
    'Tea_Stall_Owner':     {'inflow_avg': 400,   'inflow_count': 25, 'outflow_ratio': 0.80, 'stability': 0.85},
    'Flower_Vendor':       {'inflow_avg': 300,   'inflow_count': 15, 'outflow_ratio': 0.88, 'stability': 0.60}, # High volatility
    'Delivery_Partner':    {'inflow_avg': 1200,  'inflow_count': 4,  'outflow_ratio': 0.92, 'stability': 0.75}, # Bi-weekly/Weekly payout spikes
    'Driver_RideHailing':  {'inflow_avg': 1800,  'inflow_count': 6,  'outflow_ratio': 0.90, 'stability': 0.70},
    'Housekeeping_Staff':  {'inflow_avg': 8000,  'inflow_count': 1,  'outflow_ratio': 0.95, 'stability': 0.95}  # Fixed monthly chunks
}

arch_list = list(archetypes.keys())
arch_probs = [0.25, 0.20, 0.15, 0.20, 0.12, 0.08] # Distribution across India

print("Generating 10,000 localized worker profiles...")
```

```python
# 2. Vectorized Generation of User Profiles
user_ids = [f"CUST_{str(i).zfill(5)}" for i in range(1, num_workers + 1)]
assigned_archetypes = np.random.choice(arch_list, size=num_workers, p=arch_probs)
```

```python
# 3. Simulate 90-Day Transaction Matrix
# To avoid Colab memory crashes with millions of rows, we calculate 90-day aggregations directly.
# This matches the vector inputs required for your Day 4 & Day 5 ML model pipelines.
aggregated_data = []

for i in range(num_workers):
    uid = user_ids[i]
    arch = assigned_archetypes[i]
    cfg = archetypes[arch]

    # Simulate variations per individual worker
    indiv_modifier = np.random.uniform(0.7, 1.3)
    days_active = 90

    # Calculate estimated transactions over 90 days
    total_inflows_count = int(cfg['inflow_count'] * days_active * np.random.uniform(0.8, 1.2))

    # Generate financial metrics based on profile characteristics
    base_inflow = (total_inflows_count * cfg['inflow_avg'] * indiv_modifier)
    # Add random economic volatility based on their profession's stability score
    total_inflow = base_inflow * np.random.uniform(cfg['stability'], 2 - cfg['stability'])

    # Outflow calculations driven by margin constraints
    expected_outflow_ratio = cfg['outflow_ratio'] * np.random.uniform(0.95, 1.05)
    expected_outflow_ratio = min(max(expected_outflow_ratio, 0.50), 1.15) # Cap bounds
    total_outflow = total_inflow * expected_outflow_ratio

    # Extract behavioral risk features from UPI string traits
    net_savings = total_inflow - total_outflow
    savings_ratio = net_savings / total_inflow if total_inflow > 0 else 0

    # Segment transaction categories dynamically
    leisure_spend_ratio = np.random.uniform(0.02, 0.15) if savings_ratio > 0 else np.random.uniform(0.0, 0.05)
    digital_investment_affinity = np.random.uniform(0.1, 0.4) if savings_ratio > 0.10 else np.random.uniform(0.0, 0.05)

    aggregated_data.append({
        'CustomerID': uid,
        'Archetype': arch,
        'Total_Inflow_90D': round(total_inflow, 2),
        'Total_Outflow_90D': round(total_outflow, 2),
        'Net_Savings_90D': round(net_savings, 2),
        'Savings_Ratio': round(savings_ratio, 4),
        'Leisure_Spend_Ratio': round(leisure_spend_ratio, 4),
        'Investment_Affinity': round(digital_investment_affinity, 4)
    })
```

```python
# 4. Save and Inspect the Dataset
df_panel = pd.DataFrame(aggregated_data)
df_panel.to_csv('india_gigworker_90d_telemetry.csv', index=False)

print("\n--- Production Data Matrix Configured Successfully ---")
print(f"Total Rows Generated: {len(df_panel)}")
print(df_panel.groupby('Archetype')['Total_Inflow_90D'].mean().apply(lambda x: f"Avg Inflow: ₹{x:,.2f}"))
```
