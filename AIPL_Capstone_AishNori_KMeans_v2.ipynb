R Squared OLS results

```python
import statsmodels.api as sm

# Define the independent variable (X) and dependent variable (y)
X = df['Savings_Ratio']
y = df['Investment_Affinity']

# Add a constant to the independent variable for the intercept term
X = sm.add_constant(X)

# Create and fit the OLS (Ordinary Least Squares) model
model = sm.OLS(y, X)
results = model.fit()

# Print the regression summary
print(results.summary())
```

Regression Summary:

R-squared: The R-squared value is 0.409, meaning that approximately 40.9% of the variance in 'Investment_Affinity' can be explained by 'Savings_Ratio'. This indicates a moderate fit of the model to the data.
Coefficients:
The const (intercept) coefficient is -0.0181, which is the predicted 'Investment_Affinity' when 'Savings_Ratio' is zero.
The Savings_Ratio coefficient is 1.5375. This means that for every one-unit increase in 'Savings_Ratio', 'Investment_Affinity' is predicted to increase by approximately 1.5375 units.
P-values (P>|t|): Both the intercept and 'Savings_Ratio' coefficients have a p-value of 0.000, which is less than a common significance level (e.g., 0.05). This suggests that both coefficients are statistically significant, meaning they are unlikely to be zero by chance.
F-statistic & Prob (F-statistic): The model has a significant F-statistic (6933.) with a p-value of 0.00. This indicates that the overall regression model is statistically significant and 'Savings_Ratio' is a useful predictor of 'Investment_Affinity'.
The scatter plot with the regression line visually represents this relationship. The blue dots are your data points, and the red line is the ordinary least squares (OLS) regression line, which is the best-fit line through the data. Its upward slope confirms the positive relationship indicated by the positive coefficient for 'Savings_Ratio'. The shaded area around the line represents the confidence interval for the regression fit.

We have successfully performed a linear regression analysis between 'Savings_Ratio' and 'Investment_Affinity'. The analysis showed a statistically significant positive relationship, where an increase in 'Savings_Ratio' is associated with an increase in 'Investment_Affinity'. The R-squared value of 0.409 indicates that 'Savings_Ratio' explains a moderate portion of the variance in 'Investment_Affinity'. This was visualized through a scatter plot with a regression line, confirming the positive trend.

```python
# Visualize the regression results
fig = px.scatter(df, x='Savings_Ratio', y='Investment_Affinity', trendline='ols', trendline_color_override='red',
                 title='Linear Regression: Savings_Ratio vs Investment_Affinity',
                 labels={'Savings_Ratio': 'Savings Ratio', 'Investment_Affinity': 'Investment Affinity'})
fig.show()
```

```python
X = np.array(df[["Total_Inflow_90D","Total_Outflow_90D"]])
```

```python
fig = px.scatter(x=X[:,0],y=X[:,1])
fig.show()
```

K-Means Clustering with 4 cluster combination

```python
# Initialise the KMeans function and use it to find the cluster labels.

kmeans = KMeans(n_clusters=4, init='k-means++', max_iter=300, n_init=10, random_state=0)

# y is the predicted cluster label for each point in the dataset
y=kmeans.fit_predict(X)
```

```python
y
```

```python
df["label"] = y
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(name="Clusters",
      x=df["Total_Inflow_90D"],y=df["Total_Outflow_90D"],
                     mode='markers',
                     marker=dict(size=10, color=df["label"], symbol='circle')))

fig.add_trace(
    go.Scatter(name="Cluster Centers",
     x=kmeans.cluster_centers_[:,0],y=kmeans.cluster_centers_[:,1],
                     mode='markers',
                     marker=dict(size=10, color='rgb(0, 255, 0)', symbol='circle')))

fig.update_layout(xaxis_title="Total Inflow 90D",yaxis_title="Total Outflow 90D")
fig.show()
```

K-Means with 5 clusters

```python
kmeans = KMeans(n_clusters=5, init='k-means++', max_iter=300, n_init=10, random_state=0)
# y is the predicted cluster label for each point in the dataset
y=kmeans.fit_predict(X)
```

```python
y
```

```python
df["label"] = y
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(name="Clusters",
      x=df["Total_Inflow_90D"],y=df["Total_Outflow_90D"],
                     mode='markers',
                     marker=dict(size=10, color=df["label"], symbol='circle')))

fig.add_trace(
    go.Scatter(name="Cluster Centers",
     x=kmeans.cluster_centers_[:,0],y=kmeans.cluster_centers_[:,1],
                     mode='markers',
                     marker=dict(size=10, color='rgb(0, 255, 0)', symbol='circle')))

fig.update_layout(xaxis_title="Total Inflow 90D",yaxis_title="Total Outflow 90D")
fig.show()
```

WCSS - Within Cluster Sum of Square

```python
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=0)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
```

```python
fig = px.line(y=wcss)
fig.update_layout(xaxis_title="Number of Clusters",yaxis_title="WCSS")
fig.show()
```
